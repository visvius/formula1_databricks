```sql
-- This Query is used separately inside Dashboard to import dataset using SQL
WITH driver_metrics AS
(
SELECT driver_name,
    SUM(race_starts) AS race_starts,
    SUM(number_of_wins) AS total_wins,
    SUM(number_of_podiums) AS total_podiums,
    SUM(CASE WHEN standing = 1 THEN 1 ELSE 0 END) as total_championships
FROM formula1.gold.v_driver_standing
GROUP BY driver_name
HAVING total_championships >= 1
)

SELECT driver_name,
    race_starts,
    total_wins,
    total_podiums,
    total_championships,
    (total_championships * 100) + (total_wins * 10) + (total_podiums * 3) AS overall_score
FROM driver_metrics
ORDER BY overall_score DESC
    
```
